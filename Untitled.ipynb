```python
# %%
# Cell 2 — reuse robot data, expand web data, and freeze larger train/val splits

from collections import defaultdict
from concurrent.futures import ThreadPoolExecutor
from tempfile import TemporaryDirectory

from datasets import Dataset, Features, Image as HFImage, Value, load_dataset, load_from_disk
from huggingface_hub import HfApi, snapshot_download
from lerobot.common.datasets.lerobot_dataset import HF_LEROBOT_HOME
from tqdm.auto import tqdm
import torch

torch.set_num_threads(1)


def digest(payload):
    return hashlib.sha256(json.dumps(payload, sort_keys=True).encode()).hexdigest()


def atomic_json(path, payload):
    """Publish a complete JSON file, never a partially written manifest."""
    path = Path(path)
    tmp = path.with_name(path.name + ".tmp")
    tmp.write_text(json.dumps(payload, indent=2, sort_keys=True) + "\n")
    tmp.replace(path)


def freeze_json(path, payload):
    """Refuse to silently change an existing experiment's data definition."""
    if path.exists():
        if digest(json.loads(path.read_text())) != digest(payload):
            raise RuntimeError(f"Existing experiment differs: {path}. Use a new DATA_ROOT.")
    else:
        atomic_json(path, payload)


def robot_identity(root, expected_episodes):
    info_path = root / "meta/info.json"
    if not info_path.is_file():
        raise FileNotFoundError(f"Missing converted dataset metadata: {info_path}")
    info = json.loads(info_path.read_text())
    if info.get("total_episodes") != expected_episodes:
        raise RuntimeError(f"{root}: expected {expected_episodes} episodes, got {info.get('total_episodes')}")
    files = sorted(root.rglob("*.parquet")) + sorted(root.rglob("*.mp4"))
    if not any(path.suffix == ".parquet" for path in files):
        raise FileNotFoundError(f"No Parquet data under {root}")
    # Change detector, not a full byte-level integrity audit of every video.
    inventory = [(str(p.relative_to(root)), p.stat().st_size, p.stat().st_mtime_ns) for p in files]
    return {"path": str(root), "episodes": expected_episodes,
            "info_sha256": hashlib.sha256(info_path.read_bytes()).hexdigest(),
            "file_inventory_sha256": digest(inventory)}


print("[1/4] Reusing ALOHA and the existing DROID conversion...")
ALOHA_DIR = Path(snapshot_download(repo_id=ALOHA_REPO, repo_type="dataset", revision=ALOHA_REVISION))
DROID_LEROBOT_DIR = Path(HF_LEROBOT_HOME) / DROID_LEROBOT_REPO
if not (DROID_LEROBOT_DIR / "meta/info.json").is_file():
    raise FileNotFoundError(
        f"Expected the existing DROID conversion at {DROID_LEROBOT_DIR}. "
        "Correct DROID_LEROBOT_REPO/this path if it was moved. No conversion or deletion was attempted."
    )
ROBOT_IDENTITIES = {
    "aloha": robot_identity(ALOHA_DIR, 50),
    "droid": robot_identity(DROID_LEROBOT_DIR, 30),
}
for train_ids, val_ids, total in (
    (ALOHA_TRAIN_EPISODES, ALOHA_VAL_EPISODES, 50),
    (DROID_TRAIN_EPISODES, DROID_VAL_EPISODES, 30),
):
    assert set(train_ids).isdisjoint(val_ids)
    assert set(train_ids) | set(val_ids) == set(range(total))

# Resolve web revisions once. Reruns keep the same commits, even if HF main changes.
revision_path = DATA_ROOT / "web_revisions.json"
if revision_path.exists():
    WEB_REVISIONS = json.loads(revision_path.read_text())
    if set(WEB_REVISIONS) != {COCO_REPO, VQA_REPO}:
        raise RuntimeError("Web repositories changed; use a new DATA_ROOT.")
else:
    api = HfApi()
    WEB_REVISIONS = {repo: api.dataset_info(repo).sha for repo in (COCO_REPO, VQA_REPO)}
    atomic_json(revision_path, WEB_REVISIONS)

COCO_FEATURES = Features({"image": HFImage(), "filename": Value("string"),
                          "cocoid": Value("int32"), "caption": Value("string")})
VQA_FEATURES = Features({"multiple_choice_answer": Value("string"),
                         "question": Value("string"), "image": HFImage()})


def iter_web_rows(repo_id, revision, source_split, columns, unique_limit):
    """Stream encoded images to disk; never accumulate decoded images in RAM."""
    stream = load_dataset(repo_id, revision=revision, split=source_split, streaming=True)
    stream = stream.cast_column("image", HFImage(decode=False))
    seen = set()
    for row in stream:
        if unique_limit is not None:
            image_id = int(row["cocoid"])
            if image_id not in seen:
                if len(seen) >= unique_limit:
                    break
                seen.add(image_id)
        if row["image"].get("bytes") is None:
            raise RuntimeError(f"{repo_id}: expected embedded image bytes; refusing a nonportable cache.")
        yield {column: row[column] for column in columns}
    if unique_limit is not None and len(seen) < unique_limit:
        raise RuntimeError(f"{repo_id}: only found {len(seen):,} unique images.")


def acquire_web_cache(directory, repo_id, source_split, features, unique_limit=None):
    request = {"schema": 1, "repo": repo_id, "revision": WEB_REVISIONS[repo_id],
               "source_split": source_split, "unique_limit": unique_limit}
    receipt_path = directory / "acquisition.json"
    if directory.exists():
        if not (directory / "state.json").exists() or not receipt_path.exists():
            raise RuntimeError(f"Incomplete/unrecognized cache: {directory}; inspect it before retrying.")
        receipt = json.loads(receipt_path.read_text())
        if receipt["request"] != request:
            raise RuntimeError(f"Cache configuration differs: {directory}. Use a new cache directory.")
        result = load_from_disk(str(directory), keep_in_memory=False)
        if len(result) != receipt["rows"]:
            raise RuntimeError(f"Cached row count differs from its receipt: {directory}")
        return result

    # Arrow-backed random access for the later sampler. Streaming is only acquisition.
    # Builder output is kept for retry/reuse; save_to_disk creates a separate final copy.
    built = Dataset.from_generator(
        iter_web_rows,
        features=features,
        cache_dir=str(DATA_ROOT / ".builder_cache"),
        keep_in_memory=False,
        writer_batch_size=32,
        gen_kwargs={"repo_id": repo_id, "revision": WEB_REVISIONS[repo_id],
                    "source_split": source_split, "columns": list(features),
                    "unique_limit": unique_limit},
    )
    receipt = {"request": request, "rows": len(built)}
    with TemporaryDirectory(prefix=f".{directory.name}_", dir=directory.parent) as temp:
        staged = Path(temp) / "dataset"
        built.save_to_disk(str(staged), max_shard_size="256MB")
        atomic_json(staged / "acquisition.json", receipt)
        staged.rename(directory)
    return load_from_disk(str(directory), keep_in_memory=False)


print(f"[2/4] COCO: acquiring/reusing {COCO_TOTAL_IMAGES:,} unique images...")
coco = acquire_web_cache(COCO_DIR, COCO_REPO, "train", COCO_FEATURES, COCO_TOTAL_IMAGES)
coco_ids = np.asarray(coco["cocoid"], dtype=np.int64)
unique_coco_ids, first_coco_rows = np.unique(coco_ids, return_index=True)
if len(unique_coco_ids) != COCO_TOTAL_IMAGES:
    raise RuntimeError(f"Expected {COCO_TOTAL_IMAGES:,} COCO images; found {len(unique_coco_ids):,}.")
# All caption rows encountered in the selected stream prefix are retained.
# This is a bounded prefix subset, not a uniform sample of the full COCO corpus.

print("[3/4] VQAv2: acquiring/reusing the complete repository split...")
vqa = acquire_web_cache(VQA_DIR, VQA_REPO, "validation", VQA_FEATURES)
if len(vqa) <= VQA_VAL_TARGET_ROWS:
    raise RuntimeError("VQAv2 cache is too small for the requested split.")


def image_keys(dataset, indices, label):
    """Cached SHA-256 of original-size decoded RGB pixels + dimensions.

    Detects exact decoded-image duplicates, including across captioning and VQA.
    It is not a perceptual/near-duplicate image detector.
    """
    indices = np.asarray(indices, dtype=np.int64)
    identity = {"dataset": dataset._fingerprint, "algorithm": "rgb_dimensions_sha256_v1",
                "indices": hashlib.sha256(indices.astype("<i8").tobytes()).hexdigest()}
    cache_path = DATA_ROOT / f"{label}_image_keys_{digest(identity)[:16]}.npz"
    if cache_path.exists():
        with np.load(cache_path, allow_pickle=False) as stored:
            keys = stored["keys"].copy()
        if len(keys) != len(indices):
            raise RuntimeError(f"Invalid image-key cache: {cache_path}")
        return keys

    image_only = dataset.select_columns(["image"])

    def one_key(index):
        with image_only[int(index)]["image"] as image:
            with image.convert("RGB") as rgb:
                dimensions = np.asarray(rgb.size, dtype="<i4").tobytes()
                return hashlib.sha256(dimensions + rgb.tobytes()).hexdigest()

    with ThreadPoolExecutor(max_workers=DATA_SETUP_WORKERS) as pool:
        keys = np.asarray(list(tqdm(pool.map(one_key, indices), total=len(indices),
                                    desc=f"Index {label} image identities")), dtype="U64")
    tmp = cache_path.with_name(cache_path.name + ".tmp")
    with tmp.open("wb") as handle:
        np.savez_compressed(handle, keys=keys)
    tmp.replace(cache_path)
    return keys


def group_values(keys, values):
    groups = defaultdict(list)
    for key, value in zip(keys, values, strict=True):
        groups[str(key)].append(int(value))
    return dict(groups)


def validation_keys(groups, target, seed, forced_val=(), forced_train=()):
    """Choose whole image groups near a target count, honoring cross-source roles."""
    chosen, excluded = set(forced_val), set(forced_train)
    if chosen & excluded or not (chosen | excluded) <= set(groups):
        raise ValueError("Invalid/conflicting forced image groups.")
    count = sum(len(groups[key]) for key in chosen)
    candidates = sorted(set(groups) - chosen - excluded)
    np.random.default_rng(seed).shuffle(candidates)
    for key in candidates:
        if count >= target:
            break
        size = len(groups[key])
        if abs(count + size - target) <= abs(count - target):
            chosen.add(key)
            count += size
    return chosen


print("[4/4] Creating deterministic, image-grouped splits...")
coco_keys = image_keys(coco, first_coco_rows, "coco")
vqa_keys = image_keys(vqa, np.arange(len(vqa)), "vqav2")
coco_groups = group_values(coco_keys, unique_coco_ids)
vqa_groups = group_values(vqa_keys, np.arange(len(vqa)))

COCO_VAL_KEYS = validation_keys(coco_groups, COCO_VAL_IMAGES, SPLIT_SEED)
COCO_TRAIN_KEYS = set(coco_groups) - COCO_VAL_KEYS
COCO_VAL_IDS = {cid for key in COCO_VAL_KEYS for cid in coco_groups[key]}
COCO_TRAIN_IDS = set(map(int, unique_coco_ids)) - COCO_VAL_IDS
COCO_TRAIN_INDICES = np.flatnonzero(np.isin(coco_ids, sorted(COCO_TRAIN_IDS)))
COCO_VAL_INDICES = np.flatnonzero(np.isin(coco_ids, sorted(COCO_VAL_IDS)))

# merve/vqav2-small does not provide a COCO image-ID column. Use image identity,
# and keep any exact image shared with COCO on the SAME side of the split.
VQA_VAL_KEYS = validation_keys(
    vqa_groups, VQA_VAL_TARGET_ROWS, SPLIT_SEED + 1,
    forced_val=set(vqa_groups) & COCO_VAL_KEYS,
    forced_train=set(vqa_groups) & COCO_TRAIN_KEYS,
)
VQA_TRAIN_KEYS = set(vqa_groups) - VQA_VAL_KEYS
VQA_VAL_INDICES = np.flatnonzero(np.isin(vqa_keys, sorted(VQA_VAL_KEYS)))
VQA_TRAIN_INDICES = np.flatnonzero(np.isin(vqa_keys, sorted(VQA_TRAIN_KEYS)))

assert COCO_TRAIN_IDS.isdisjoint(COCO_VAL_IDS)
assert (COCO_TRAIN_KEYS | VQA_TRAIN_KEYS).isdisjoint(COCO_VAL_KEYS | VQA_VAL_KEYS)
for train_rows, val_rows, total in (
    (COCO_TRAIN_INDICES, COCO_VAL_INDICES, len(coco)),
    (VQA_TRAIN_INDICES, VQA_VAL_INDICES, len(vqa)),
):
    assert len(train_rows) > 0 and len(val_rows) > 0
    assert not np.intersect1d(train_rows, val_rows).size
    assert len(train_rows) + len(val_rows) == total

# The robot statistics filename depends on the actual train episodes and data identity.
# No old statistics are loaded or computed in these first two cells.
ROBOT_NORM_SPEC = {
    "schema": 1, "quantiles": [0.01, 0.99], "action_horizon": ACTION_HORIZON,
    "aloha": {"data": ROBOT_IDENTITIES["aloha"], "train_episodes": ALOHA_TRAIN_EPISODES,
              "native_dim": 14, "actions": "PI convention; 6 joint deltas + gripper per arm"},
    "droid": {"data": ROBOT_IDENTITIES["droid"], "train_episodes": DROID_TRAIN_EPISODES,
              "native_dim": 8, "actions": "7 joint velocities + gripper position"},
}
ROBOT_NORM_ID = digest(ROBOT_NORM_SPEC)[:16]
NORM_STATS_PATH = DATA_ROOT / f"robot_quantile_stats_{ROBOT_NORM_ID}.npz"
NORM_METADATA_PATH = NORM_STATS_PATH.with_suffix(".json")
# Legacy DROID stats came from SORTED RAW paths; converted episode indices need
# not have that order. Cell 3 must verify provenance before reusing those values,
# or recompute from converted training episodes 0..23 (no video decoding needed).

SPLIT_SPEC = {
    "schema": 1, "split_seed": SPLIT_SEED, "source_probabilities": MIXTURE_WEIGHTS,
    "aloha": {"data": ROBOT_IDENTITIES["aloha"], "repo": ALOHA_REPO, "revision": ALOHA_REVISION,
              "train_episodes": ALOHA_TRAIN_EPISODES, "val_episodes": ALOHA_VAL_EPISODES},
    "droid": {"data": ROBOT_IDENTITIES["droid"], "repo": DROID_LEROBOT_REPO,
              "train_episodes": DROID_TRAIN_EPISODES, "val_episodes": DROID_VAL_EPISODES},
    "coco": {"repo": COCO_REPO, "revision": WEB_REVISIONS[COCO_REPO], "path": str(COCO_DIR),
             "fingerprint": coco._fingerprint, "rows": len(coco),
             "train_image_ids": sorted(COCO_TRAIN_IDS), "val_image_ids": sorted(COCO_VAL_IDS),
             "train_indices": COCO_TRAIN_INDICES.tolist(), "val_indices": COCO_VAL_INDICES.tolist()},
    "vqav2": {"repo": VQA_REPO, "revision": WEB_REVISIONS[VQA_REPO], "path": str(VQA_DIR),
              "source_split": "validation", "fingerprint": vqa._fingerprint, "rows": len(vqa),
              "train_indices": VQA_TRAIN_INDICES.tolist(), "val_indices": VQA_VAL_INDICES.tolist()},
    "image_grouping": "COCO IDs + exact original-size decoded RGB SHA-256; cross-source roles aligned",
}
SPLIT_ID = digest(SPLIT_SPEC)[:16]
manifest = {"split_id": SPLIT_ID, "splits": SPLIT_SPEC,
            "robot_normalization": ROBOT_NORM_SPEC, "normalization_path": str(NORM_STATS_PATH)}
freeze_json(MANIFEST_PATH, manifest)

print("\nSource       Train      Validation   Unit")
print(f"ALOHA        {len(ALOHA_TRAIN_EPISODES):>7,}    {len(ALOHA_VAL_EPISODES):>7,}      episodes")
print(f"DROID        {len(DROID_TRAIN_EPISODES):>7,}    {len(DROID_VAL_EPISODES):>7,}      episodes")
print(f"COCO         {len(COCO_TRAIN_IDS):>7,}    {len(COCO_VAL_IDS):>7,}      unique image IDs")
print(f"VQAv2        {len(VQA_TRAIN_INDICES):>7,}    {len(VQA_VAL_INDICES):>7,}      question/answer rows")
print(f"COCO caption rows cached: {len(coco):,}; VQAv2 rows cached: {len(vqa):,}")
print(f"Exact image identities shared by COCO/VQA: {len(set(coco_groups) & set(vqa_groups)):,}")
print("Source probabilities:", MIXTURE_WEIGHTS)
print("Split ID:", SPLIT_ID)
print("Manifest:", MANIFEST_PATH)
print("New normalization destination:", NORM_STATS_PATH)
print("Cell 2 ready. Stop here; normalization is the next cell.")
```
